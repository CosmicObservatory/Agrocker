```python
df = pd.DataFrame(columns=['Estado', 'Pais', 'Latitud', 'Longitud', "Fecha"]+ ['maxtemp_c',
 'mintemp_c',
 'avgtemp_c',
 'totalprecip_mm',
 'totalsnow_cm',
 'avgvis_km',
 'avghumidity',
 'uv'])

def jsontodf(data_js):
    vars = ['maxtemp_c',
            'mintemp_c',
            'avgtemp_c',
            'totalprecip_mm',
            'totalsnow_cm',
            'avgvis_km',
            'avghumidity',
            'uv']
    locate = data_js['location']
    climate = data_js['forecast']['forecastday']
    locate_data = [locate['region'], locate['country'], locate['lat'], locate['lon']]
    data = []
    for day in climate:
        day_date = [day['date']]
        climate_date = day['day']
        climate_data = [climate_date[k] for k in vars]
        weather_data = locate_data + day_date + climate_data
        data.append(weather_data)
    return data

jsontodf(data)
```
